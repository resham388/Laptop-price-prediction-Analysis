```python
plt.figure(figsize = (10,8))
sns.boxplot(x = data['OS'], y= data['Price_euros'])
plt.title('OS VS CPU_freqency')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.barplot(
    x=data['Touchscreen'],
    y=data['Price_euros'],
    hue=data['Screen'],
    palette='Set2'  # Options: 'Set1', 'Set2', 'Set3', 'pastel', 'muted', 'husl'
)
plt.title('Touchscreen vs Price by Screen Type')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.barplot(
    x=data['OS'],
    y=data['Price_euros'],
    hue=data['Touchscreen'],
    palette='Set1'  )
plt.title('OS vs Price by Touchscreen')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.barplot(
    x=data['PrimaryStorage'],
    y=data['Price_euros'],
    hue=data['SecondaryStorage'],
    palette='Set2' 
)
plt.title('Primary vs Secondary Storage and Price')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.barplot(
    x=data['RetinaDisplay'],
    y=data['Price_euros'],
    hue=data['PrimaryStorage'],
    palette='Set2' 
)
plt.title('Retina Display vs Price by Primary Storage')
plt.show()
```

```python
sns.scatterplot(
    data=data,
    x='CPU_freq',
    y='Ram',
    hue='OS',
    palette='Set2'  
)
plt.title('CPU Frequency vs RAM by OS')
plt.show()
```

```python
sns.scatterplot(
    data=data,
    x='CPU_freq',
    y='Ram',
    hue='OS',
    palette='Set1'  # or 'Set2', 'Set3', 'husl'
)
plt.title('CPU Frequency vs RAM by OS')
plt.show()
```
